```python
s = ''.join(input().split())
def sol(s):
    if '*' in s:
        m = list(map(float, s.split('*')))
        return m[0]*m[1]
    if '/' in s:
        m = list(map(float, s.split('/')))
        return m[0]/m[1]
    if '+' in s:
        m = list(map(float, s.split('+')))
        return m[0]+m[1]
    if '-' in s:
        m = list(map(float, s.split('-')))
        return m[0]-m[1]
print(round(sol(s), 2))
```

```python
a, b = map(int, (input().split()))
def sol(a, b):
    c = a^b
    s=0
    while c!=0:
        s+=c%2
        c//=2
    return s
print(sol(a, b))
```

```python
a, b = map(str, input().split())
kt = {}
def sol():
    global a, b, kt
    for i in range(len(a)):
        if a[i] not in kt: 
            kt[a[i]] = b[i]
        elif b[i]!=kt[a[i]]: 
            return False
    for i in kt:
        for j in kt:
            if i!=j:
                if kt[i]==kt[j]: 
                    return False
    return True
print(sol())
```
