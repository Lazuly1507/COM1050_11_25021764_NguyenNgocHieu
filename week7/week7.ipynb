```python
l = list(map(int, input().split()))
tar = int(input().strip())
le = 0
ri = len(l)-1
def search(arr, le, ri, tar):
    while le <= ri:
        mid_arr = (le + ri)//2
        if tar == arr[mid_arr]: 
            return mid_arr
        if tar < arr[mid_arr]:
            ri = mid_arr - 1
        if tar > arr[mid_arr]:
            le = mid_arr + 1
    return -1
print(search(l, le, ri, tar))
```

```python
l = list(map(int, input().split()))
def bbsort(l):
    n = len(l)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1, n):
            if l[j] < l[min_idx]:
                min_idx = j
        if min_idx != i:
            l[i], l[min_idx] = l[min_idx], l[i]
bbsort(l)
print(l)
```

```python
import ast
arr =  ast.literal_eval(input())
def msort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2
    left = arr[:mid]
    right = arr[mid:]
    left = msort(left)
    right = msort(right)
    return merge(left, right)

def merge(left, right):
    res = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i]>=right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res

ans = []
for i in range(len(arr)):
    if i%2==1:
        ans.extend(msort(arr[i]))
    else:
        ans.extend(arr[i])
print(ans)
```

```python
import ast
arr = ast.literal_eval(input())
def bsearch(arr, tar):
    left = 0
    right = len(arr)-1
    while left<=right:
        mid = (left + right)//2
        if tar == arr[mid]:
            return mid
        if tar < arr[mid]:
            right = mid-1
        else:
            left = mid + 1
    return left
    
minx = []
for i in range(len(arr)):
    j = bsearch(minx, arr[i])
    if j == len(minx):
        minx.append(arr[i])
    else:
        minx[j] = arr[i]
print(len(minx))
```

```python
import ast
arr = ast.literal_eval(input())
que = ast.literal_eval(input())
left = 0
right = len(arr)-1
def ubs(arr, left, right, tar):
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == tar:
            return mid
        if arr[mid] < tar:
            left = mid + 1
        else:
            right = mid -1
    return left

def lbs(arr, left, right, tar):
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == tar:
            return mid
        if arr[mid] < tar:
            left = mid + 1
        else:
            right = mid -1
    return right

minx = [-1]*len(que)
for i in range(len(arr)):
    i = arr[i][1] - arr[i][0] + 1
    le = ubs(que, 0, len(que)-1, arr[i][0])
    ri = lbs(que, 0, len(que)-1, arr[i][1])

    for j in range(le, ri+1):
        if minx[j] == -1 or minx[j] > i:
            minx[j] = i

print(minx)
```
