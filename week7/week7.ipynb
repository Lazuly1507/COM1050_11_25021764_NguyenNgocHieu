```python
l = list(map(int, input().split()))
tar = int(input().strip())
le = 0
ri = len(l)-1
def search(arr, le, ri, tar):
    while le <= ri:
        mid_arr = (le + ri)//2
        if tar == arr[mid_arr]: 
            return mid_arr
        if tar < arr[mid_arr]:
            ri = mid_arr - 1
        if tar > arr[mid_arr]:
            le = mid_arr + 1
    return -1
print(search(l, le, ri, tar))
```

```python
l = list(map(int, input().split()))
def bbsort(l):
    n = len(l)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1, n):
            if l[j] < l[min_idx]:
                min_idx = j
        if min_idx != i:
            l[i], l[min_idx] = l[min_idx], l[i]
            
bbsort(l)
print(l)
```

```python
import ast
arr =  ast.literal_eval(input())
def msort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2
    left = arr[:mid]
    right = arr[mid:]
    left = msort(left)
    right = msort(right)
    return merge(left, right)

def merge(left, right):
    res = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i]>=right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res

ans = []
for i in range(len(arr)):
    if i%2==1:
        ans.extend(msort(arr[i]))
    else:
        ans.extend(arr[i])
print(ans)
```

```python
import ast
arr = ast.literal_eval(input())
def bsearch(arr, tar):
    left = 0
    right = len(arr)-1
    while left<=right:
        mid = (left + right)//2
        if tar == arr[mid]:
            return mid
        if tar < arr[mid]:
            right = mid-1
        else:
            left = mid + 1
    return left
    
minx = []
for i in range(len(arr)):
    j = bsearch(minx, arr[i])
    if j == len(minx):
        minx.append(arr[i])
    else:
        minx[j] = arr[i]
print(len(minx))
```

```python
def bseact(arr, tar):
    left = 0
    right = len(arr)-1
    while left <= right:
        mid = (left + right)//2
        if arr[mid][0] == tar:
            return mid
        if arr[mid][0] <= tar:
            left = mid + 1
        else:
            right = mid-1
    return right

def Min(a, b):
    if a == -1:
        return b
    else:
        return min(a, b)
import ast
arr = ast.literal_eval(input())
que = ast.literal_eval(input())
n = len(que)
minx = [-1]*n

for i in range(n):
    idx = bseact(arr, que[i])
    if idx != -1:
        for j in range(idx+1):
            if que[i] <= arr[j][1]:
                l = arr[j][1] - arr[j][0] + 1
                minx[i] = Min(minx[i], l)
        
print(minx)
```

```python
#A10
import ast
arr = ast.literal_eval(input())
m = int(input())

def ubs(arr, tar):
    left = 0
    right = len(arr)-1
    while left<=right:
        mid = (left+right)//2
        if arr[mid]>tar:
            right=mid-1
        elif arr[mid]<=tar:
            left=mid+1
    return left
Sum = [0]
cur_sum = 0
ans =0 

for i in range(0, len(arr)):
    #print(Sum, cur_sum)
    cur_sum = (cur_sum + arr[i])%m
    ans = max(cur_sum, ans)
    idx = ubs(Sum, cur_sum)
    if idx < len(Sum):
        ans = max((cur_sum - Sum[idx])%m, ans)
    Sum.insert(idx, cur_sum)

print(ans)
```
