```python
#A1

def merge(left, right):
    res = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res
    
a = list(map(int, input().split()))
b = list(map(int, input().split()))

ans = merge(a, b)

for i in ans:
    print(i, end = " ")
```

```python
#A2

def input_matrix(mt = []):
    a = list(map(int, input().split()))
    if len(a) >1:
        res = merge(mt, a)
        return input_matrix(res)
    else:
        return mt, a[0]

ds, k = input_matrix()
print(ds[k-1])
```

```python
#A4

def sol(a, b):
    i, j = 0, 0
    while i<len(a) and j<len(b):
        if a[i].upper() !=b[j]:
            if a[i].isupper():
                return "NO"
        else:
            j+=1
        i+=1
    if j >= len(b):
        return "YES"
    return "NO"

Q = int(input().strip())

for i in range(Q):
    a = input().strip()
    b = input().strip()
    print(sol(a, b))
```
