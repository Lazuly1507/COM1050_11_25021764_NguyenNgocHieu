```python
#A1

def merge(left, right):
    res = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res
    
a = list(map(int, input().split()))
b = list(map(int, input().split()))

ans = merge(a, b)

for i in ans:
    print(i, end = " ")
```

```python
#A2

def input_matrix(mt = []):
    a = list(map(int, input().split()))
    if len(a) >1:
        res = merge(mt, a)
        return input_matrix(res)
    else:
        return mt, a[0]

ds, k = input_matrix()
print(ds[k-1])
```

```python
#A3
def find_smallest_swap(s_str):
    s = list(s_str)
    n = len(s)
    
    for i in range(n - 1):
        min_idx = i
        for j in range(n - 1, i, -1):
            if s[j] < s[min_idx]:
                if min_idx == i or s[j] < s[min_idx]:
                    min_idx = j
        
        if min_idx != i:
            s[i], s[min_idx] = s[min_idx], s[i]
            return "".join(s)
            
    s[-1], s[-2] = s[-2], s[-1]
    return "".join(s)
```

```python
def solve():
    import sys
    input = sys.stdin.read
    data = input().split()
    
    Q = int(data[0])
    idx = 1
    
    results = []
    for _ in range(Q):
        a = data[idx]
        b = data[idx+1]
        idx += 2
        
        n, m = len(a), len(b)
        dp = [[False] * (m + 1) for _ in range(n + 1)]
        
        dp[0][0] = True
        
        for i in range(1, n + 1):
            for j in range(m + 1):
                if dp[i-1][j]:
                    if a[i-1].islower():
                        dp[i][j] = True
                
                if j > 0 and dp[i-1][j-1]:
                    if a[i-1].upper() == b[j-1]:
                        dp[i][j] = True
        
        if dp[n][m]:
            results.append("YES")
        else:
            results.append("NO")
            
    print('\n'.join(results))
```

```python
def check_sublist(list1, list2):
    n1 = len(list1)
    n2 = len(list2)

    if n1 == 0:
        return True
    
    for i in range(n2 - n1 + 1):
        match = True
        
        for j in range(n1):
            if list2[i + j] != list1[j]:
                match = False 
                break
        
        if match == True:
            return True
            
    return False

# Chạy thử
a = [2, 3, 4]
b = [1, 2, 3, 4, 5]

if check_sublist(a, b):
    print("YES")
else:
    print("NO")
```
