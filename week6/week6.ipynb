W6A4

```python
a = list(map(str, input().split()))
b = []
ans = {}
for i in a:
    b+=i.split(':')
i=0
while i<len(b)-1:
    if b[i] not in ans:
        ans[b[i]]=[b[i+1]]
    else:
        ans[b[i]].append(b[i+1])
    i+=2
print(ans)
```

```python
a = list(map(int, input().split()))
ans = {'positives':0, 'negatives': 0, 'zeros':0}
for i in a:
    if i>0: ans['positives']+=1
    elif i<0: ans['negatives']+=1
    else: ans['zeros']+=1
print(ans)
```
