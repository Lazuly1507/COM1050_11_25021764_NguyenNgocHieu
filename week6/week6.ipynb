```python
import ast
a = ast.literal_eval(input())
b = ast.literal_eval(input())
cnt = {}
for i in a:
    cnt[i] = a[i]
for i in b:
    if i in a:
        cnt[i]+=b[i]
    else: 
        cnt[i]=b[i]
cnt = list(cnt.items())
cnt.sort()
cnt = dict(cnt)
print(cnt)
```

```python
a = list(map(int, input().split()))
cnt = {}
for i in a:
    if i in cnt:
        cnt[i]+=1
    else:
        cnt[i] = 1
maxn=max(cnt.values())
ans = min(x for x in cnt if maxn==cnt[x])
print(ans)
```
