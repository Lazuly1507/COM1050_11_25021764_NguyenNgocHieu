```python
#A1
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
        self.sd = 3.14*r*r
        self.sxq = 2*r*3.14*h
        self.stp = 2*self.sd+self.sxq
        self.v = self.sd*h
    def __repr__(self):
        return f"{self.stp:.2f} {self.v:.2f}" 
r, h = map(int, input().split())
cylinder = Cylinder(r, h)
print(cylinder)
```

```python
class Date:
    m30 = [4, 6, 9, 11]
    m31 = [1, 3, 5, 7, 8, 10, 12]
    def __init__(self, d, m, y):
        self.d = d
        self.m = m
        self.y = y
        self.leap = self.check_year()
    def check_year(self):
        if self.y%400==0:
            return True
        if self.y % 100==0:
            return False
        if self.y%4==0:
            return True
        return False
    def check_valid(self):
        if self.d<=0:
            return False
        if self.m<=0:
            return False
        if self.y<=0:
            return False
        if self.m<=12:
            if self.m == 2:
                if self.leap and self.d > 29:
                    return False
                elif not self.leap and self.d > 28:
                    return False
            else:
                if self.m in self.m30:
                    if self.d >30:
                        return False
                elif self.m in self.m31:
                    if self.d >31:
                        return False
            return True
        else:
            return False
    def is_end_of_month(self):
        if self.m == 2:
            return self.d == 29 if self.leap else self.d == 28
        if self.m in self.m30:
            return self.d == 30
        return self.d == 31
    def next_day(self):
        check = self.check_valid()
        if check:
            if self.is_end_of_month():
                self.d=1
                if self.m == 12:
                    self.m=1
                    self.y+=1
                else:
                    self.m+=1
            else:
                self.d+=1
            return True
        else:
            return False
    def __repr__(self):
        return f"{self.d:02}/{self.m:02}/{self.y:04}"

d, m, y = map(int, input().split('/'))
date = Date(d, m, y)
if date.next_day():
    print(date)
else:
    print('INVALID')
```

```python
#A3
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def is_axis(self):
        if self.x== self.y==0:
            return "M la goc toa do"
        if self.x==0:
            return "M nam tren truc tung"
        if self.y==0:
            return "M nam tren truc hoanh"
    def dist(self, other):
        dx = self.x-other.x
        dy = self.y-other.y
        return ((dx*dx+dy*dy)**0.5)

x, y = map(float, input().split())
M = Point(x, y)
O = Point(0, 0)
if M.is_axis():
    print(M.is_axis())
print(f"Khoang cach tu M den goc toa do la {M.dist(O):.2f}")
```
