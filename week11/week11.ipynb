```python
a, b = map(int, input().split())
try:
    print(f"{a/b:.2f}")
except ZeroDivisionError:
    print("Loi chia khong")
```

```python
a=list(map(int, input().split()))
n=len(a)
try:
    print(a[n])
except IndexError as e:
    print(e)
```

```python
a=input()
def sol(a):
    sum=0
    for i in a:
        sum+=(int(i)**2)
    if sum==1:
        return "YES"
    return sol(str(sum))
try:
    print(sol(a))
except RecursionError:
    print("NO")
except ValueError as e:
    print(e)
```
